## Bonus (if time remains) — one line, zero loops, every patch at once

Everything above generalizes into something you'll want for Assignment 1:
`numpy.lib.stride_tricks.sliding_window_view` extracts **every** sliding
window of a given shape from an array, all at once, with no loop at all.

Run the cell below (no TODOs — just look at the shapes) and think about how
this connects to what you just practiced: it's built entirely out of the
same indexing and broadcasting ideas.

````python
from numpy.lib.stride_tricks import sliding_window_view

patches = sliding_window_view(grid, (3, 3))
print("patches shape:", patches.shape)   # (6, 6, 3, 3): one 3x3 window per valid center
print("patches[0, 0]:\n", patches[0, 0])  # the first 3x3 window, top-left corner
print("patches[2, 3]:\n", patches[2, 3])  # the window centered around row 3, col 4

# If `kernel` were a 3x3 filter, applying it to every window at once is now
# a single broadcasting expression -- no pixel loop required:
kernel = np.ones((3, 3)) / 9
filtered = (patches * kernel).sum(axis=(-1, -2))
print("filtered shape:", filtered.shape)
````
